```python
""""Change _trips.tntp file to _od.csv"""

network_name = "Munich" #"Philadelphia" #Birmingham. GoldCoast

import re

def parse_data_line(line):
    entries = re.findall(r'(\d+)\s*:\s*([\d\.]+);', line)
    return [(int(dest), float(value)) for dest, value in entries]


input_filename = '../data/{}/{}_trips.tntp'.format(network_name, network_name)
output_filename = '../data/{}/{}_od.csv'.format(network_name, network_name)


with open(input_filename, 'r') as file:
    lines = file.readlines()

current_origin = None
data = []


for line in lines:
    if 'Origin' in line:
        # Extract the origin number
        current_origin = int(re.search(r'Origin\s+(\d+)', line).group(1))
    else:
        # Parse the data line and append the origin
        entries = parse_data_line(line)
        
        for destination, value in entries:
            data.append([current_origin, destination, value])


# Write to CSV
with open(output_filename, 'w') as file:
    file.write("O,D,Ton\n")
    for origin, destination, value in data:
        file.write(f"{origin},{destination},{value}\n")


```

```python
# For Winnipeg
import re

network_name = "Barcelona"  # Change to your desired network
input_filename = f'../data/{network_name}/{network_name}_trips.tntp'
output_filename = f'../data/{network_name}/{network_name}_od.csv'

def parse_data_line(line):
    return re.findall(r'(\d+)\s*:\s*([\d\.]+)', line)

data = []
current_origin = None

with open(input_filename, 'r') as file:
    lines = file.readlines()

for line in lines:
    line = line.strip()
    if not line:
        continue

    if line.startswith("Origin"):
        match = re.match(r"Origin\s+(\d+)", line)
        if match:
            current_origin = int(match.group(1))
    elif current_origin is not None:
        for dest, val in parse_data_line(line):
            data.append([current_origin, int(dest), float(val)])

# Write to CSV
with open(output_filename, 'w') as file:
    file.write("O,D,Ton\n")
    for origin, destination, value in data:
        file.write(f"{origin},{destination},{value}\n")
```
